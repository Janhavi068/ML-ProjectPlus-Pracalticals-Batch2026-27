```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import fetch_california_housing
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

np.random.seed(42)
X_simple = np.linspace(0, 10, 100).reshape(-1, 1)
y_simple = (2.5 * X_simple.ravel() + 5.0 + np.random.normal(0, 2.0, size=100))

X_b = np.c_[np.ones((len(X_simple), 1)), X_simple]  # Add bias column (x0 = 1)
theta_numpy = np.linalg.pinv(X_b.T @ X_b) @ X_b.T @ y_simple
y_pred_numpy = X_b @ theta_numpy

rmse_numpy = np.sqrt(mean_squared_error(y_simple, y_pred_numpy))
r2_numpy = r2_score(y_simple, y_pred_numpy)
print(f"[Simple LR - NumPy]        Intercept: {theta_numpy[0]:.4f}, Slope: {theta_numpy[1]:.4f}, RMSE: {rmse_numpy:.4f}, R²: {r2_numpy:.4f}")

lr_simple = LinearRegression()
lr_simple.fit(X_simple, y_simple)
y_pred_sk_simple = lr_simple.predict(X_simple)

rmse_sk_simple = np.sqrt(mean_squared_error(y_simple, y_pred_sk_simple))
r2_sk_simple = r2_score(y_simple, y_pred_sk_simple)
print(f"[Simple LR - scikit-learn] Intercept: {lr_simple.intercept_:.4f}, Slope: {lr_simple.coef_[0]:.4f}, RMSE: {rmse_sk_simple:.4f}, R²: {r2_sk_simple:.4f}")

residuals_simple = y_simple - y_pred_sk_simple
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].scatter(X_simple, y_simple, color="steelblue", alpha=0.7, label="Actual Data")
axes[0].plot(X_simple, y_pred_sk_simple, color="crimson", linewidth=2.2, label="Fitted Regression Line")
axes[0].set_title("Simple Linear Regression: Fitted Line")
axes[0].set_xlabel("Feature (X)")
axes[0].set_ylabel("Target (y)")
axes[0].legend()
axes[0].grid(True, linestyle="--", alpha=0.4)

axes[1].scatter(y_pred_sk_simple, residuals_simple, color="darkorange", alpha=0.7)
axes[1].axhline(0, color="black", linestyle="--", linewidth=1.5)
axes[1].set_title("Simple Linear Regression: Residuals vs. Predicted")
axes[1].set_xlabel("Predicted Values (ŷ)")
axes[1].set_ylabel("Residuals (y - ŷ)")
axes[1].grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()

data = fetch_california_housing()
X_multi, y_multi = data.data, data.target

X_train, X_test, y_train, y_test = train_test_split(
    X_multi, y_multi, test_size=0.2, random_state=42
)

# --- 2a. Using NumPy (Normal Equation with bias term) ---
X_train_b = np.c_[np.ones((X_train.shape[0], 1)), X_train]
X_test_b = np.c_[np.ones((X_test.shape[0], 1)), X_test]

theta_multi = np.linalg.pinv(X_train_b.T @ X_train_b) @ X_train_b.T @ y_train
y_test_pred_numpy = X_test_b @ theta_multi

rmse_multi_np = np.sqrt(mean_squared_error(y_test, y_test_pred_numpy))
r2_multi_np = r2_score(y_test, y_test_pred_numpy)
print(f"\n[Multiple LR - NumPy]        Test RMSE: {rmse_multi_np:.4f}, Test R²: {r2_multi_np:.4f}")

# --- 2b. Using scikit-learn ---
lr_multi = LinearRegression()
lr_multi.fit(X_train, y_train)
y_test_pred_sk = lr_multi.predict(X_test)

rmse_multi_sk = np.sqrt(mean_squared_error(y_test, y_test_pred_sk))
r2_multi_sk = r2_score(y_test, y_test_pred_sk)
print(f"[Multiple LR - scikit-learn] Test RMSE: {rmse_multi_sk:.4f}, Test R²: {r2_multi_sk:.4f}")

# Plot Predicted vs Actual and Residuals for Multiple Linear Regression
residuals_multi = y_test - y_test_pred_sk

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].scatter(y_test, y_test_pred_sk, alpha=0.4, color="teal")
min_val = min(y_test.min(), y_test_pred_sk.min())
max_val = max(y_test.max(), y_test_pred_sk.max())
axes[0].plot([min_val, max_val], [min_val, max_val], color="crimson", lw=2, label="Ideal Fit (y = ŷ)")
axes[0].set_title("Multiple Linear Regression: Actual vs. Predicted")
axes[0].set_xlabel("Actual Target")
axes[0].set_ylabel("Predicted Target")
axes[0].legend()
axes[0].grid(True, linestyle="--", alpha=0.4)

axes[1].scatter(y_test_pred_sk, residuals_multi, alpha=0.4, color="purple")
axes[1].axhline(0, color="black", linestyle="--", linewidth=1.5)
axes[1].set_title("Multiple Linear Regression: Residuals vs. Predicted")
axes[1].set_xlabel("Predicted Values (ŷ)")
axes[1].set_ylabel("Residuals (y - ŷ)")
axes[1].grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```
